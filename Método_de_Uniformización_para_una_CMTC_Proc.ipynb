### **Solución**

```python
import numpy as np
import sympy as sp
import math
```

Se importan las librerias necesarias para los cálculos

```python
R = sp.Matrix([[0, 2, 3, 0],
               [4, 0, 2, 0],
               [0, 2, 0, 2],
               [1, 0, 3, 0]])
```

Se tiene la matriz R proporcionada

```python
def matP(R):
  n = R.shape[0]
  P = sp.zeros(n,n)
  l = []
  for i in range(n):
    r = sum(R[i,:])
    l.append(r)
    r = max(l)
  for i in range(n):
    for j in range(n):
      if i == j:
        P[i,j] = 1 - sum(R[i,:])/r
      else:
        P[i,j] = R[i,j]/r
  return P, r
```

Se crea una función donde dada la matriz R, construye la matriz P a partir de la teoría.

```python
P, r = matP(R)
P
```

Se tiene la matriz P, que es exactamente igual a la que se obtuvo en el ejericicio 1.

```python
def p(t,R):
  n = R.shape[0]
  P, r = matP(R)
  M = max(int(r*t + 5 * sp.sqrt(r*t)), 20)
  s = sp.zeros(n, n)
  for i in range(M):
    s = s + ((r*t)**i)/math.factorial(i) * P**i
  pt = sp.exp(-r*t) * s

  return pt, M
```

Se crea una función donde a partir de la matriz R se obtiene

$P(t) = \sum_{k=0}^{\infty} e^{-rt}\,\frac{(rt)^k}{k!}\,\hat{P}^{\,k}$

```python
t=0.5
H, M = p(t,R)
H
```

$P(0.5)$ se tiene la matriz cuando $t=0.5$

```python
t=1.0
T = p(t,R)
T
```

$P(1)$

```python
t=5.0
T2 = p(t,R)
T2
```

$P(5)$

```python
H*H
```

Nótese que se cumplen las ecuaciones de Champan Kolmogorov, ya que:

$P(0.5)P(0.5) = P(1)$

# **Ejercicio 4.2**

Suponga que se desea calcular $P(t)$ con una tolerancia $\varepsilon$. Elija $M$ tal que

$$\sum_{k=M+1}^{\infty} e^{-rt} \frac{(rt)^k}{k!} \le \varepsilon$$

Y se puede implementar de acuerdo al siguiente algoritmo de uniformización para $P(t)$:

1. Dados $R$, $t$, $0 < \varepsilon < 1$.
2. Calcular $r$ usando la igualdad en la definición.
3. Calcular $\hat{P}$.
4. $A = \hat{P}$; $B = e^{-rt} I$; $c = e^{-rt}$; $sum = c$; $k = 1$

5. Mientras $sum < 1 - \varepsilon$ hacer:

$$c = c \cdot \frac{rt}{k}$$
$$B = B + cA$$
$$A = A \hat{P}$$
$$sum = sum + c$$
$$k = k + 1$$

6. $B$ está a $\varepsilon$ de $P(t)$.

Repita el ejercicio 3 aplicando este algoritmo con una tolerancia $ \varepsilon = 0.00001 $ (indique el valor correspondiente de M en cada caso). Compare los resultados.

### **Solución**

---

En el ejercicio 3, lo valores de $M$ son:

Para $ t=0.5$, $ M=20$

Para $ t=1$, $ M=20$

Para $ t=5$, $ M=57$

```python
def uni(R,t,e):
  P, r = matP(R)
  A = P
  B = sp.exp(-r*t)*sp.eye(R.shape[0])
  c = sp.exp(-r*t)
  s = c
  k = 1

  while s < 1 - e:
    c = c * r * t/k
    B = B + c*A
    A = A*P
    s = s + c
    k += 1
  return B, k
```

Se crea una función a partir del algoritmo dado

##### **$P(0.5)$**

```python
t=0.5
e=0.00001
P = uni(R,t,e)
P
```

En este caso $M=14$, que es menor comparado con el $M$ del ejercicio 3

##### **$P(1)$**

```python
t=1.0
e=0.00001
P1 = uni(R,t,e)
P1
```

En este caso $M=20$, que coincide con el $M$ del ejercicio 3

##### **$P(5)$**

```python
t=5.0
e=0.00001
P2 = uni(R,t,e)
P2
```

En este caso $M=57$, que coincide con el $M$ del ejercicio 3

En general se tienen resultado muy similares, pero en este último ejercicio se calcula hasta cierta tolerancia $\varepsilon$
